## O que é uma Série Temporal?

Uma série temporal é qualquer conjunto de observações em que **cada ponto de dado está associado a um instante de tempo específico**. Isso significa que, ao citar um valor, você sempre o referencia a uma data ou horário — por exemplo, *"o preço da ação era 200 nesta data"*.

---

## Exemplos de Séries Temporais

| Dado | Por quê é uma série temporal? |
|------|-------------------------------|
| **Preço de ações** (ex.: Apple via Yahoo Finance) | Cada preço é registrado em uma data/hora específica |
| **Precipitação (rainfall)** | Medições diárias ou horárias de chuva |
| **Clima (weather)** | Temperatura, pressão etc. registrados ao longo do tempo |


---

## Diferença entre Séries Temporais e Dados Estáticos (Cross-Sectional)

| Aspecto | Série Temporal | Dados Estáticos/Cross-Sectional |
|---------|---------------|----------------------------------|
| **Indexação** | Por tempo | Por observação (sem ordem temporal) |
| **Dependência temporal** | Presente e fundamental | Ausente |
| **Exemplo** | Preço diário de ações | Detecção de fraude em cartão de crédito |

Na detecção de fraude, por exemplo, o que importa são as *features* da transação (local, valor, titular), **não** quando ela aconteceu. Já em séries temporais, a **ordem e a posição no tempo** são a essência do problema.

---

## Atenção: Data Leakage em Séries Temporais

Por causa da dependência temporal, é essencial **sempre ter o fator tempo em mente**. Usar acidentalmente dados que ocorreram *depois* do ponto de investigação contamina o modelo — isso é chamado de **vazamento de dados (data leakage)**. Esse erro leva a resultados enganosos e modelos que não generalizam para o futuro.

---

## Forecasting (Previsão)

**Forecasting** é a forma de predição específica para séries temporais. Diferente de modelos supervisionados clássicos, o objetivo aqui é explicitamente **prever valores futuros** com base nos valores passados observados.

> *Exemplo:* "O preço da ação da Apple foi X hoje. Qual será o preço amanhã?"

Enquanto modelos de classificação/regressão tradicionais identificam padrões em features estáticas, o forecasting explora a **estrutura temporal** dos dados para projetar o que ainda não aconteceu.

```python
import yfinance as yf
import plotly.graph_objects as go


# Define the stock and time period:
ticker_symbol = 'AAPL'
start_date = '2021-01-01'
end_date = '2023-01-01'

# Get the data from Yahoo Finance:
data = yf.download(ticker_symbol, start=start_date, end=end_date)
```

```python
data.head()
```

```python
# Create the Plot:
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=data.index,
    y=data[('Close', ticker_symbol)],   # acessa corretamente o MultiIndex
    mode='lines',
    name='Close Price'
))
fig.update_layout(
    title=f'{ticker_symbol} Stock Prices from Yahoo Finance',
    xaxis_title='Date',
    yaxis_title='Price (USD)',
    template='simple_white',
    width=1000,    # largura razoável
    height=500,
)
fig.show()
```
